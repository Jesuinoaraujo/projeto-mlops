```python
import pandas as pd

# Caminho relativo a partir do diretório do notebook
bronze_path = '../data/bronze/'
df1 = pd.read_csv(f'{bronze_path}nafld1.csv')
df2 = pd.read_csv(f'{bronze_path}nafld2.csv')

# Exibição inicial dos dados
print("NAFLD1:")
display(df1.head())

print("NAFLD2:")
display(df2.head())

# Verificação da estrutura das colunas
print("Colunas NAFLD1:", df1.columns)
print("Colunas NAFLD2:", df2.columns)

# Junção dos DataFrames
df_merged = pd.concat([df1, df2], axis=0, ignore_index=True)

# Exibição dos dados mesclados
print("DataFrame Mesclado:")
display(df_merged.head())

# Tratamento de valores ausentes (exemplo)
df_merged.fillna(method='ffill', inplace=True)

# Salvando o DataFrame como Parquet na camada Silver
silver_path = '../data/silver/'
df_merged.to_parquet(f'{silver_path}nafld_merged.parquet', engine='pyarrow')

print("Transformação concluída e arquivo salvo na camada Silver como Parquet.")
```
