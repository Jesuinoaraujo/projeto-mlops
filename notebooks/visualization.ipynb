```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caminho para os dados Silver
silver_path = '../data/silver/nafld_merged.parquet'

# Leitura do arquivo Parquet
df = pd.read_parquet(silver_path, engine='pyarrow')

# Histograma do BMI
plt.figure(figsize=(10, 6))
sns.histplot(df['bmi'], kde=True, bins=30)
plt.title('Distribuição do BMI')
plt.xlabel('BMI')
plt.ylabel('Frequência')
plt.show()

# Gráfico de dispersão entre BMI e Idade
plt.figure(figsize=(10, 6))
sns.scatterplot(x='age', y='bmi', data=df)
plt.title('Relação entre Idade e BMI')
plt.xlabel('Idade')
plt.ylabel('BMI')
plt.show()
```
